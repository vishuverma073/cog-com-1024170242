**b) Find the most frequent value and its indices.**

```python
def most_frequent(a):
    """Print the most frequent value(s) of an array and where they occur."""
    values, counts = np.unique(a, return_counts=True)
    highest = counts.max()
    winners = values[counts == highest]   # more than one value can share the top count

    print("Array :", a)
    print("Counts:", dict(zip(values.tolist(), counts.tolist())))
    for value in winners:
        print("Most frequent value:", value, "occurs", highest, "times at indices",
              np.where(a == value)[0])
    print()
```

```python
x = np.array([1, 2, 3, 4, 5, 1, 2, 1, 1, 1])
most_frequent(x)

y = np.array([1, 1, 1, 2, 3, 4, 2, 4, 3, 3])
most_frequent(y)   # here 1 and 3 both appear 3 times, so both are printed
```

## Question 4

Create a 1-D array named `vishu` with 25 evenly spaced numbers from 10 to 100 using `linspace()`.

```python
vishu = np.linspace(10, 100, 25)
print(vishu)
```

```python
print("Dimensions (ndim) :", vishu.ndim)
print("Shape             :", vishu.shape)
print("Total elements    :", vishu.size)
print("Data type         :", vishu.dtype)
print("Bytes per element :", vishu.itemsize)
print("Total bytes       :", vishu.nbytes)
```

```python
# transpose using reshape -- turn the row of 25 into a column of 25
transposed = vishu.reshape(25, 1)
print("Shape after reshape(25, 1):", transposed.shape)
print(transposed[:5], "...")
```

```python
# can we do the same with T?
print("Shape of vishu.T:", vishu.T.shape)
print("Is vishu.T the same as vishu?", np.array_equal(vishu.T, vishu))
```

**Answer:** No, `T` cannot do it here. Transposing swaps the axes, and a 1-D array has only one
axis, so `vishu.T` gives back the very same 25-element row. To get a column you have to add a
second axis, which is what `reshape(25, 1)` does (`vishu[:, np.newaxis]` works too). `T` only
transposes properly once the array is 2-D.

## Question 5

Create a 2-D array of three rows and four columns named `ucs420_vishu`.

```python
ucs420_vishu = np.array([[10, 20, 30, 40],
                         [50, 60, 70, 80],
                         [90, 15, 20, 35]])
print(ucs420_vishu)
print("Shape:", ucs420_vishu.shape)
```

```python
print("Mean            :", ucs420_vishu.mean())
print("Median          :", np.median(ucs420_vishu))
print("Max             :", ucs420_vishu.max())
print("Min             :", ucs420_vishu.min())
print("Unique elements :", np.unique(ucs420_vishu))
```

```python
reshaped_ucs420_vishu = ucs420_vishu.reshape(4, 3)
print("reshaped_ucs420_vishu (4 rows, 3 columns):")
print(reshaped_ucs420_vishu)
```

```python
resized_ucs420_vishu = np.resize(ucs420_vishu, (2, 3))
print("resized_ucs420_vishu (2 rows, 3 columns):")
print(resized_ucs420_vishu)

# reshape needs the element count to match (12 = 4x3), resize does not --
# 2x3 only holds 6 values, so resize keeps the first 6 and drops the rest
print("\nElements kept:", resized_ucs420_vishu.size, "out of", ucs420_vishu.size)
```
