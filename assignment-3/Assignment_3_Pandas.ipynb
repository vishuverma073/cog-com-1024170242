## Question 6Create the sample dataset `employees.csv` with information about employees in a company.

```python
employees = pd.DataFrame({
    "Employee_ID": [101, 102, 103, 104, 105],
    "Name": ["Alice", "Bob", "Charlie", "Diana", "Edward"],
    "Department": ["HR", "IT", "IT", "Marketing", "Sales"],
    "Age": [29, 34, 41, 28, 38],
    "Salary": [50000, 70000, 65000, 55000, 60000],
    "Years_of_Experience": [4, 8, 10, 3, 12],
    "Joining_Date": ["2020-03-15", "2017-07-19", "2013-06-01", "2021-02-10", "2010-11-25"],
    "Gender": ["Female", "Male", "Male", "Female", "Male"],
    "Bonus": [5000, 7000, 6000, 4500, 5000],
    "Rating": [4.5, 4.0, 3.8, 4.7, 3.5],
})

employees.to_csv("employees.csv", index=False)
employees = pd.read_csv("employees.csv")
employees
```

**a) Shape of the DataFrame.**

```python
print("Shape:", employees.shape)
print("Rows:", employees.shape[0], "Columns:", employees.shape[1])
```

**b) Summary with data types and non-null counts.**

```python
employees.info()
```

**c) Descriptive statistics.**

```python
employees.describe()
```

**d) First 5 rows and last 3 rows.**

```python
print("First 5 rows:")
display(employees.head())

print("Last 3 rows:")
display(employees.tail(3))
```

**e) Statistics from the dataset.**

```python
print("i.   Average salary        :", employees["Salary"].mean())
print("ii.  Total bonus paid      :", employees["Bonus"].sum())
print("iii. Youngest employee age :", employees["Age"].min())
print("iv.  Highest rating        :", employees["Rating"].max())
```

**f) Sort the DataFrame by Salary in descending order.**

```python
employees.sort_values(by="Salary", ascending=False)
```

**g) Add a column categorising employees on their performance rating.**

```python
def performance(rating):
    if rating >= 4.5:
        return "Excellent"
    elif rating >= 4.0:
        return "Good"
    else:
        return "Average"

employees["Performance"] = employees["Rating"].apply(performance)
employees[["Name", "Rating", "Performance"]]
```

**h) Identify missing values.**

```python
print("Missing values in each column:")
print(employees.isnull().sum())
print("\nTotal missing values:", employees.isnull().sum().sum())
```

**i) Rename Employee_ID to ID.**

```python
employees = employees.rename(columns={"Employee_ID": "ID"})
employees.head()
```

**j) Find employees with more than 5 years of experience, and employees in IT.**

```python
print("i. More than 5 years of experience:")
display(employees[employees["Years_of_Experience"] > 5])

print("ii. Employees in the IT department:")
display(employees[employees["Department"] == "IT"])
```

**k) Add a Tax column that deducts 10% of the Salary.**

```python
employees["Tax"] = employees["Salary"] * 0.10
employees[["Name", "Salary", "Tax"]]
```

**l) Save the modified DataFrame to a new CSV file.**

```python
employees.to_csv("employees_modified.csv", index=False)
print("Saved to employees_modified.csv")

# read it back to confirm the new columns are there
pd.read_csv("employees_modified.csv")
```
