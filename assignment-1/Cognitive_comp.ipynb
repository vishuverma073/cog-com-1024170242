## Question 1

```python
roll_number = "1024170242"
L = [int(digit) * 10 for digit in roll_number]
print("Original list L:", L)
```

```python
print("List L:", L)
```

```python
L.append(55)
print("After append(55):", L)   # 55 is added right at the end of the list

L.insert(3, 99)
print("After insert(3, 99):", L)   # 99 takes index 3 and everything after it shifts one place right
```

```python
L.remove(70)
print("After remove(70):", L)

popped = L.pop(2)
print("After pop(2):", L)
print("Value removed by pop:", popped)
```

```python
L.sort()
print("Ascending:", L)

L.sort(reverse=True)
print("Descending:", L)
```

```python
print("First three:", L[:3])
print("Last three:", L[-3:])
```

```python
average = sum(L) / len(L)
above_average = [n for n in L if n > average]
print("Average of L:", average)
print("Greater than average:", above_average)
```

## Question 2

```python
original_L = [int(digit) * 10 for digit in roll_number]   # L was edited and sorted in Q1, so rebuilding it here
scores = tuple(original_L[:8])
print("scores =", scores)
```

```python
highest = max(scores)
lowest = min(scores)
print("Highest score:", highest, "at index", scores.index(highest))
print("Lowest score:", lowest, "appears", scores.count(lowest), "time(s)")
```

```python
# a tuple is immutable so it has no .reverse() of its own, reversed() builds a new sequence instead
print("Reversed as list:", list(reversed(scores)))
```

```python
value = int(input("Enter a score: "))
if value in scores:
    print(value, "first occurs at index", scores.index(value))
else:
    print(value, "is not present in scores")
```

```python
try:
    scores[0] = 100
except TypeError as err:
    print("Error raised:", err)
    # tuples block item assignment completely, the same line on a list would have worked because lists are mutable
```

```python
first, second, *rest = scores
print("First:", first)
print("Second:", second)
print("Rest:", rest)
```

## Question 3

```python
import random
from collections import Counter

random.seed(int(roll_number))
numbers = [random.randint(100, 900) for _ in range(100)]
print("100 random numbers:", numbers)
```

```python
odd = [n for n in numbers if n % 2 != 0]
print("Odd count:", len(odd))
print("Odd numbers:", odd)
```

```python
even = [n for n in numbers if n % 2 == 0]
print("Even count:", len(even))
print("Even numbers:", even)
```

```python
def is_prime(n):
    if n < 2:
        return False
    for i in range(2, int(n ** 0.5) + 1):
        if n % i == 0:
            return False
    return True


primes = [n for n in numbers if is_prime(n)]
print("Prime count:", len(primes))
print("Prime numbers:", primes)
```

```python
number, times = Counter(numbers).most_common(1)[0]
print("Most frequent number:", number)
print("It occurs", times, "times")
```

## Question 4

```python
digits = [int(digit) for digit in roll_number][:8]   # the roll number has 10 digits, the question asks for 8
A = {d * 7 for d in digits}
B = {d * 9 for d in digits}
print("Digits used:", digits)
print("A =", A)
print("B =", B)
```

```python
print("Union:", A.union(B))
```

```python
print("Intersection:", A.intersection(B))
```

```python
print("A - B:", A.difference(B))
print("B - A:", B.difference(A))
# difference() is one sided and keeps only what the left set has, symmetric_difference() collects the leftovers of both sides at once
```

```python
print("Symmetric difference:", A.symmetric_difference(B))
```

```python
print("Is A a subset of B:", A.issubset(B))
print("Is B a superset of A:", B.issuperset(A))
```

```python
x = int(input("Enter a value to remove from A: "))
A.discard(x)   # discard() quietly does nothing when x is missing, remove() would raise a KeyError
print("A after discard:", A)
```

## Question 5

```python
my_dict = {
    "name": "Vishu Verma",
    "roll_no": "1024170242",
    "branch": "CSE",
    "age": 20,
    "city": "Ludhiana",
}
print(my_dict)
```

```python
my_dict["location"] = my_dict.pop("city")
print("After renaming city to location:", my_dict)
```

```python
my_dict["cgpa"] = 8.4
print("After adding cgpa:", my_dict)
```

```python
my_dict["age"] += 1
print("After increasing age by 1:", my_dict)
```

```python
copy_for_pop = dict(my_dict)
removed = copy_for_pop.pop("branch")
print("pop() returned:", removed)
print("Copy after pop:", copy_for_pop)

copy_for_del = dict(my_dict)
del copy_for_del["branch"]
print("Copy after del:", copy_for_del)
# pop() hands the removed value back so it can still be used, del only deletes the key and returns nothing
```

```python
for key, value in my_dict.items():
    print(f"{key} -> {value}")
```

```python
if "email" in my_dict:
    print("Email:", my_dict["email"])
else:
    print("Email is not stored in this dictionary")
```

```python
friend_dict = {
    "name": "Aman Gupta",
    "roll_no": "1024170199",
    "branch": "ECE",
    "age": 21,
    "city": "Jaipur",
}

merged = {**my_dict, **friend_dict}
print("Merged:", merged)
# the dictionary unpacked last wins a shared key, so friend_dict overwrites my_dict
```

```python
only_strings = {key: value for key, value in my_dict.items() if isinstance(value, str)}
print("String values only:", only_strings)
```
